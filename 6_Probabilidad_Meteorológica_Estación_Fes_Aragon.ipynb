# Probabilidad Meteorológica Estación FES Aragón

```python
import pandas as pd
```

```python
path = "/content/meteorologia_2025.csv"
df_temp = pd.read_csv(
    path,
    sep=',',
    skiprows=9,
    header=0,
    low_memory=False
)

df = df_temp.rename(columns={'valor': 'value'}).drop(columns=['unit'])
df.columns = ['datetime', 'station_id', 'variable', 'value']
print("Filas cargadas:", len(df))
```

```python
df['value'] = pd.to_numeric(df['value'], errors='coerce')

df_pivot = df.pivot_table(
    index=['datetime', 'station_id'],
    columns='variable',
    values='value',
    aggfunc='first'
).reset_index()

df_pivot.columns.name = None

display(df_pivot.head())
```

```python
df_estacion = df_pivot[df_pivot["station_id"] == 'FAR'].copy()
df_estacion = df_estacion.dropna(subset=["RH", "WSP"])

total_registros = len(df_estacion)
print(f"Registros válidos: {total_registros}")
display(df_estacion.head())
```

```python
cond_hum_alta = df_estacion["RH"] >= 70
cond_viento_tranq = df_estacion["WSP"] <= 1
```

a) Probabilidad de humedad alta ($ \ge 70$)

```python
p_hum_alta = cond_hum_alta.sum() / total_registros
```

b) Probabilidad de viento tranquilo ($ \le 1$)

```python
p_viento_tranq = cond_viento_tranq.sum() / total_registros
```

c) $P(\text{humedad_alta} \mid \text{viento_tranquilo})$

```python
ambos = (cond_hum_alta & cond_viento_tranq).sum()
p_hum_dado_viento = ambos / cond_viento_tranq.sum()
```

d) $P(\text{viento_tranquilo} \mid \text{humedad_alta})$

```python
p_viento_dado_hum = ambos / cond_hum_alta.sum()
```

```python
print(f"a) P(Humedad Alta): {p_hum_alta:.4f}")
print(f"b) P(Viento Tranquilo): {p_viento_tranq:.4f}")
print(f"c) P(Humedad Alta | Viento Tranquilo): {p_hum_dado_viento:.4f}")
print(f"d) P(Viento Tranquilo | Humedad Alta): {p_viento_dado_hum:.4f}")
```
